```python
# Celda 0: Importar librerias
import os
from datetime import datetime
```

```python
# Celda 1: Configurar Rutas

# Nombres de tus scripts y archivos CSV
HISTORICO_SCRIPT = "historical.py" # Tu script original para actualizar el histórico (antes results.py)
SUCCESSES_SCRIPT = "successes.py"
NEW_GAME_SCRIPT = "new_game.py"

HISTORICO_CSV = "baloto_historico_completo.csv"
MIS_JUEGOS_GENERADOS_CSV = "mis_juegos_generados.csv"

print("Orquestación de la automatización de Baloto iniciada.")
print(f"Archivos de script esperados: {HISTORICO_SCRIPT}, {SUCCESSES_SCRIPT}, {NEW_GAME_SCRIPT}")
print(f"Archivos de datos esperados: {HISTORICO_CSV}, {MIS_JUEGOS_GENERADOS_CSV}")

# Verificar si los scripts existen
for script in [HISTORICO_SCRIPT, SUCCESSES_SCRIPT, NEW_GAME_SCRIPT]:
    if not os.path.exists(script):
        print(f"⚠️ ¡Advertencia! El script '{script}' no se encontró en la misma carpeta.")
        print("Asegúrate de que todos los archivos .py estén en la misma ubicación que este notebook.")
```

```python
# Celda 4: Generar un Nuevo Número para el Próximo Juego

print("\n--- PASO 3: Generando un nuevo número para jugar ---")

# Pedir la fecha directamente en la celda de Jupyter
fecha_valida = False
fecha_juego_str = ""
while not fecha_valida:
    fecha_juego_str = input("Ingresa la fecha para la que generas el juego (DD-MM-AAAA): ")
    try:
        # Una validación simple de formato (puedes añadir más robustez si lo deseas)
        datetime.strptime(fecha_juego_str, '%d-%m-%Y') 
        fecha_valida = True
    except ValueError:
        print("Formato de fecha inválido. Por favor, usa DD-MM-AAAA.")

# Ahora, ejecuta el script pasándole la fecha como argumento
print(f"Ejecutando: python {NEW_GAME_SCRIPT} {fecha_juego_str}")
!python {NEW_GAME_SCRIPT} {fecha_juego_str}

print("\nGeneración del nuevo juego completada. Puedes encontrarlo en la primera línea de 'mis_juegos_generados.csv'.")
```
