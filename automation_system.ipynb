```python
# ============================================================================
# CELDA 1: CONFIGURACIÓN Y VERIFICACIÓN (ACTUALIZADA CON NUEVA ESTRUCTURA)
# ============================================================================

print("=" * 70)
print("🎰 SISTEMA DE BALOTO - VERSIÓN ORGANIZADA")
print("=" * 70)

import os
import sys
import json
import pandas as pd
import subprocess
from datetime import datetime

print("✅ Librerías importadas")

# ============================================================================
# CONFIGURACIÓN DE RUTAS (NUEVA ESTRUCTURA)
# ============================================================================

# Definir rutas según estructura organizada
RUTAS = {
    # Carpetas
    'scripts': 'scripts/',
    'data': 'data/',
    'reports': 'reports/',
    
    # Scripts Python
    'historical': 'scripts/historical.py',
    'analyzer': 'scripts/analyzer.py', 
    'new_game': 'scripts/new_game.py',
    'successes': 'scripts/successes.py',
    'learning': 'scripts/learning_system_enhanced.py',
    
    # Archivos CSV (datos)
    'historico_csv': 'data/baloto_historico_completo.csv',
    'mis_juegos_csv': 'data/mis_juegos_generados.csv',
    
    # Archivos JSON (reportes)
    'analisis_json': 'reports/analisis_baloto.json',
    'knowledge_json': 'reports/baloto_knowledge.json',
    'recomendaciones_json': 'reports/recomendaciones_actuales.json',
    'performance_report': 'reports/learning_performance_report.json'
}

# Verificar que existan las carpetas
print("\n📁 VERIFICANDO ESTRUCTURA DE CARPETAS...")
carpetas_requeridas = ['scripts', 'data', 'reports']
for carpeta in carpetas_requeridas:
    if os.path.exists(carpeta):
        print(f"   ✅ {carpeta}/")
    else:
        print(f"   ⚠️  {carpeta}/ (no existe - se creará automáticamente)")
        os.makedirs(carpeta, exist_ok=True)

# Función para mover archivos a nuevas ubicaciones
def reorganizar_archivos_existentes():
    """Mueve archivos existentes a la nueva estructura organizada"""
    print("\n🔄 ORGANIZANDO ARCHIVOS EXISTENTES...")
    
    # Mapeo de archivos a mover
    archivos_a_mover = {
        # Scripts Python → scripts/
        'historical.py': RUTAS['historical'],
        'analyzer.py': RUTAS['analyzer'],
        'new_game.py': RUTAS['new_game'],
        'successes.py': RUTAS['successes'],
        'learning_system_enhanced.py': RUTAS['learning'],
        
        # Archivos CSV → data/
        'baloto_historico_completo.csv': RUTAS['historico_csv'],
        'mis_juegos_generados.csv': RUTAS['mis_juegos_csv'],
        
        # Archivos JSON → reports/
        'analisis_baloto.json': RUTAS['analisis_json'],
        'baloto_knowledge.json': RUTAS['knowledge_json'],
        'recomendaciones_actuales.json': RUTAS['recomendaciones_json'],
        'learning_performance_report.json': RUTAS['performance_report']
    }
    
    movidos = 0
    for origen, destino in archivos_a_mover.items():
        if os.path.exists(origen) and not os.path.exists(destino):
            try:
                # Crear directorio destino si no existe
                directorio_destino = os.path.dirname(destino)
                if directorio_destino and not os.path.exists(directorio_destino):
                    os.makedirs(directorio_destino, exist_ok=True)
                
                # Mover archivo
                os.rename(origen, destino)
                print(f"   📦 {origen} → {destino}")
                movidos += 1
            except Exception as e:
                print(f"   ❌ Error moviendo {origen}: {e}")
        elif os.path.exists(destino):
            print(f"   ✅ {os.path.basename(destino)} ya está en ubicación correcta")
    
    if movidos > 0:
        print(f"\n✅ {movidos} archivos movidos a la nueva estructura")
    else:
        print(f"\n📋 Todos los archivos ya están organizados")

# Ejecutar reorganización
reorganizar_archivos_existentes()

# Verificar archivos esenciales en nuevas ubicaciones
print("\n🔍 VERIFICANDO ARCHIVOS ESENCIALES (nuevas ubicaciones):")

archivos_esenciales = [
    ("historical.py", RUTAS['historical']),
    ("analyzer.py", RUTAS['analyzer']), 
    ("new_game.py", RUTAS['new_game']),
    ("successes.py", RUTAS['successes']),
    ("baloto_historico_completo.csv", RUTAS['historico_csv'])
]

todos_presentes = True
for nombre, ruta in archivos_esenciales:
    if os.path.exists(ruta):
        print(f"   ✅ {nombre} ({ruta})")
    else:
        print(f"   ❌ {nombre} (FALTANTE en {ruta})")
        todos_presentes = False

# Verificar otros archivos importantes (no críticos)
print("\n🔍 VERIFICANDO OTROS ARCHIVOS:")
archivos_secundarios = [
    ("learning_system_enhanced.py", RUTAS['learning']),
    ("analisis_baloto.json", RUTAS['analisis_json']),
    ("baloto_knowledge.json", RUTAS['knowledge_json'])
]

for nombre, ruta in archivos_secundarios:
    if os.path.exists(ruta):
        print(f"   ✅ {nombre}")
    else:
        print(f"   ⚠️  {nombre} (se generará automáticamente)")

# Mostrar resumen de estructura
print("\n📁 ESTRUCTURA DEL PROYECTO:")
print("   scripts/      → Archivos Python (.py)")
print("   data/         → Archivos de datos (.csv)")
print("   reports/      → Archivos de reportes (.json)")
print("   automation_system.ipynb → Sistema principal")

if not todos_presentes:
    print("\n⚠️  ADVERTENCIA: Faltan archivos esenciales")
    print("   Algunas funcionalidades pueden no estar disponibles")
    print("   Asegúrate de que los scripts estén en scripts/")
else:
    print("\n✅ Sistema listo para usar")

print("\n" + "=" * 70)
```

```python
# ============================================================================
# CELDA 2: ACTUALIZAR DATOS HISTÓRICOS (AUTOMÁTICO) - ACTUALIZADA
# ============================================================================

print("🔄 ACTUALIZAR DATOS DESDE LA WEB - MODO AUTOMÁTICO")
print("=" * 70)

def actualizar_datos_automatico():
    """Ejecuta historical.py automáticamente, solo si es necesario"""
    if not os.path.exists(RUTAS['historical']):
        print(f"❌ historical.py no encontrado en: {RUTAS['historical']}")
        return False
    
    # Verificar si ya tenemos datos actualizados hoy
    necesita_actualizar = True
    
    if os.path.exists(RUTAS['historico_csv']):
        try:
            # Obtener fecha de modificación del archivo
            tiempo_modificacion = os.path.getmtime(RUTAS['historico_csv'])
            fecha_modificacion = datetime.fromtimestamp(tiempo_modificacion)
            hoy = datetime.now()
            
            # Si se modificó hoy, no actualizar
            if fecha_modificacion.date() == hoy.date():
                print("📅 Los datos ya están actualizados hoy")
                print(f"   Última actualización: {fecha_modificacion.strftime('%Y-%m-%d %H:%M:%S')}")
                print(f"   Ubicación: {RUTAS['historico_csv']}")
                necesita_actualizar = False
                
                # Verificar si el archivo tiene contenido
                tamano = os.path.getsize(RUTAS['historico_csv'])
                if tamano < 1024:  # Menos de 1KB, probablemente vacío
                    print("⚠️  Archivo muy pequeño, forzando actualización...")
                    necesita_actualizar = True
                    
        except Exception as e:
            print(f"⚠️  Error verificando fecha: {e}")
            # Continuar con actualización por seguridad
            necesita_actualizar = True
    
    if not necesita_actualizar:
        print("✅ Datos ya actualizados hoy, omitiendo descarga")
        return True
    
    print("📥 Descargando datos actualizados...")
    print(f"   Script: {RUTAS['historical']}")
    print(f"   Destino: {RUTAS['historico_csv']}")
    
    try:
        resultado = subprocess.run(
            [sys.executable, RUTAS['historical']],
            capture_output=True,
            text=True,
            encoding='utf-8',
            timeout=120  # 2 minutos máximo
        )
        
        if resultado.returncode == 0:
            print("✅ Datos actualizados correctamente")
            
            # Mostrar estadísticas clave del output
            output = resultado.stdout
            
            # Buscar información relevante
            lineas_interesantes = [
                "sorteos NUEVOS encontrados",
                "Total de sorteos",
                "Sorteos agregados",
                "Actualización completada"
            ]
            
            info_encontrada = False
            for linea in output.split('\n'):
                linea_lower = linea.lower()
                if any(termino in linea_lower for termino in ["sorteo", "actualiz", "nuevo", "total"]):
                    if not linea.startswith("Warning") and not linea.startswith("Deprecation"):
                        print(f"   📊 {linea.strip()}")
                        info_encontrada = True
            
            if not info_encontrada:
                # Mostrar últimas líneas significativas
                lineas = output.strip().split('\n')
                if len(lineas) > 3:
                    for linea in lineas[-3:]:
                        if linea.strip():
                            print(f"   {linea.strip()}")
            
            # Verificar que el archivo se creó/actualizó
            if os.path.exists(RUTAS['historico_csv']):
                tamano = os.path.getsize(RUTAS['historico_csv'])
                if tamano > 1024:  # Más de 1KB
                    print(f"💾 Archivo guardado: {RUTAS['historico_csv']} ({tamano:,} bytes)")
                else:
                    print(f"⚠️  Archivo muy pequeño, puede haber errores: {RUTAS['historico_csv']}")
            
            return True
            
        else:
            print("❌ Error actualizando datos")
            
            # Mostrar error específico si existe
            if resultado.stderr:
                errores = resultado.stderr.split('\n')
                for error in errores[:5]:  # Mostrar solo primeros 5 errores
                    if error.strip() and not error.startswith("Warning"):
                        print(f"   ⚠️  {error.strip()}")
            
            # Verificar si al menos tenemos el archivo
            if os.path.exists(RUTAS['historico_csv']):
                print(f"📁 Usando datos existentes (a pesar del error): {RUTAS['historico_csv']}")
                return True
            else:
                return False
                
    except subprocess.TimeoutExpired:
        print("⏰ Tiempo excedido. La descarga está tomando demasiado tiempo.")
        print("📁 Usando datos existentes si están disponibles...")
        
        if os.path.exists(RUTAS['historico_csv']):
            print(f"✅ Usando datos existentes: {RUTAS['historico_csv']}")
            return True
        else:
            print(f"❌ No hay datos existentes disponibles en: {RUTAS['historico_csv']}")
            return False
            
    except Exception as e:
        print(f"❌ Error inesperado: {e}")
        return False

# Ejecutar actualización automática
print("\n🔍 Verificando estado de datos...")
actualizar_datos_automatico()

print("\n" + "=" * 70)
```

```python
# ============================================================================
# CELDA 3: ANÁLISIS ESTADÍSTICO (AUTOMÁTICO) - ACTUALIZADA
# ============================================================================

print("📊 EJECUTAR ANÁLISIS ESTADÍSTICO - MODO AUTOMÁTICO")
print("=" * 70)

print(f"🔍 Python que se usará: {sys.executable}")
print(f"📂 Está en .venv?: {'.venv' in sys.executable}")

def ejecutar_analisis_automatico():
    """Ejecuta analyzer.py automáticamente para análisis completo"""
    if not os.path.exists(RUTAS['analyzer']):
        print(f"❌ analyzer.py no encontrado en: {RUTAS['analyzer']}")
        return False
    
    # Verificar si existen los datos necesarios
    if not os.path.exists(RUTAS['historico_csv']):
        print("❌ No hay datos históricos para analizar")
        print(f"   Ubicación esperada: {RUTAS['historico_csv']}")
        print("   Ejecuta primero la Celda 2 para obtener datos")
        return False
    
    print("🔍 Iniciando análisis estadístico automático...")
    print(f"   Datos: {RUTAS['historico_csv']}")
    print(f"   Script: {RUTAS['analyzer']}")
    
    # Verificar tamaño del archivo de datos
    try:
        tamano_datos = os.path.getsize(RUTAS['historico_csv'])
        if tamano_datos < 1024:  # Menos de 1KB
            print(f"⚠️  Archivo de datos muy pequeño ({tamano_datos} bytes), puede estar vacío o corrupto")
            return False
    except Exception as e:
        print(f"⚠️  Error verificando tamaño del archivo: {e}")
        return False
    
    try:
        # Ejecutar analyzer.py con opciones completas
        resultado = subprocess.run(
            [sys.executable, RUTAS['analyzer'], "--modo", "completo", "--salida", "ambos"],
            capture_output=True,
            text=True,
            encoding='utf-8',
            timeout=60  # 1 minuto máximo
        )
        
        if resultado.returncode == 0:
            print("✅ Análisis completado exitosamente")
            
            # Extraer y mostrar información clave del output
            output = resultado.stdout
            lineas = output.split('\n')
            
            # Buscar secciones importantes
            secciones_importantes = [
                "REPORTE DE ANÁLISIS",
                "TOP 10 NÚMEROS MÁS FRECUENTES",
                "TOP 10 NÚMEROS MENOS FRECUENTES", 
                "SUPERBALOTAS MÁS FRECUENTES",
                "RECOMENDACIONES:",
                "ANÁLISIS COMPLETADO"
            ]
            
            print("\n📋 RESUMEN DEL ANÁLISIS:")
            imprimiendo = False
            
            for linea in lineas:
                linea_stripped = linea.strip()
                
                # Iniciar impresión cuando encontramos sección importante
                if any(seccion in linea for seccion in secciones_importantes):
                    imprimiendo = True
                    if linea_stripped:
                        print(f"\n{linea_stripped}")
                        print("-" * min(50, len(linea_stripped)))
                elif imprimiendo and linea_stripped:
                    # Solo imprimir líneas con contenido (no espacios)
                    if any(marcador in linea for marcador in ["•", "Número", "Superbalota", "Estrategia", "Números"]):
                        print(f"   {linea_stripped}")
                elif linea_stripped == "":
                    # Línea vacía, continuar
                    continue
                elif "="*60 in linea:
                    # Fin de sección
                    imprimiendo = False
            
            # Mostrar información del archivo generado
            if os.path.exists(RUTAS['analisis_json']):
                tamano_analisis = os.path.getsize(RUTAS['analisis_json'])
                print(f"\n💾 Resultados guardados en: {RUTAS['analisis_json']} ({tamano_analisis:,} bytes)")
                
                # Cargar y mostrar datos clave
                try:
                    with open(RUTAS['analisis_json'], "r", encoding='utf-8') as f:
                        datos = json.load(f)
                    
                    if 'basico' in datos:
                        basico = datos['basico']
                        print(f"📈 DATOS BÁSICOS:")
                        if 'total_baloto' in basico:
                            print(f"   • Sorteos analizados: {basico['total_baloto']:,}")
                        if 'fecha_inicio' in basico and 'fecha_fin' in basico:
                            print(f"   • Período: {basico['fecha_inicio']} a {basico['fecha_fin']}")
                    
                    if 'recomendaciones' in datos:
                        rec = datos['recomendaciones']
                        print(f"\n🎯 RECOMENDACIONES CLAVE:")
                        if 'estrategia_sugerida' in rec:
                            print(f"   • Estrategia: {rec['estrategia_sugerida'].upper()}")
                        if 'numeros_calientes' in rec and rec['numeros_calientes']:
                            print(f"   • Números calientes: {', '.join(map(str, rec['numeros_calientes'][:5]))}")
                        if 'numeros_frios' in rec and rec['numeros_frios']:
                            print(f"   • Números fríos: {', '.join(map(str, rec['numeros_frios'][:3]))}")
                        if 'superbalotas_sugeridas' in rec and rec['superbalotas_sugeridas']:
                            print(f"   • Superbalotas: {', '.join(map(str, rec['superbalotas_sugeridas']))}")
                            
                except json.JSONDecodeError as e:
                    print(f"⚠️  Error leyendo archivo JSON: {e}")
                except Exception as e:
                    print(f"⚠️  Error procesando resultados: {e}")
            
            return True
            
        else:
            print("❌ Error durante el análisis")
            
            # Mostrar errores específicos
            if resultado.stderr:
                errores = [e for e in resultado.stderr.split('\n') if e.strip()]
                if errores:
                    print("\n⚠️  Errores encontrados:")
                    for error in errores[:3]:  # Mostrar solo primeros 3 errores
                        print(f"   • {error}")
            
            # Verificar si al menos se generó el archivo
            if os.path.exists(RUTAS['analisis_json']):
                print(f"📁 Usando análisis existente (a pesar de errores): {RUTAS['analisis_json']}")
                return True
            else:
                return False
                
    except subprocess.TimeoutExpired:
        print("⏰ Tiempo excedido en análisis")
        print("📁 Usando análisis existente si está disponible...")
        
        if os.path.exists(RUTAS['analisis_json']):
            print(f"✅ Usando análisis existente: {RUTAS['analisis_json']}")
            
            # Mostrar fecha de modificación
            try:
                tiempo_mod = os.path.getmtime(RUTAS['analisis_json'])
                fecha_mod = datetime.fromtimestamp(tiempo_mod)
                print(f"   Generado: {fecha_mod.strftime('%Y-%m-%d %H:%M:%S')}")
            except:
                pass
            
            return True
        else:
            print(f"❌ No hay análisis previo disponible en: {RUTAS['analisis_json']}")
            return False
            
    except Exception as e:
        print(f"❌ Error inesperado: {e}")
        return False

# Ejecutar análisis automático
print("\n🔍 Verificando datos para análisis...")
ejecutar_analisis_automatico()

print("\n" + "=" * 70)
```

```python
# ============================================================================
# CELDA 4: SISTEMA DE APRENDIZAJE MEJORADO (AUTOMÁTICO) - ACTUALIZADA
# ============================================================================

print("🧠 SISTEMA DE APRENDIZAJE AUTOMÁTICO - EJECUCIÓN AUTOMÁTICA")
print("=" * 70)

def ejecutar_aprendizaje_automatico():
    """Ejecuta el sistema de aprendizaje automáticamente"""
    
    # Verificar si existe el archivo de aprendizaje (USANDO RUTAS)
    if not os.path.exists(RUTAS['learning']):
        print(f"⚠️  learning_system_enhanced.py no encontrado")
        print(f"   Ruta esperada: {RUTAS['learning']}")
        print("   Creando archivo básico de aprendizaje...")
        crear_archivo_aprendizaje_basico()
    
    # Verificar requisitos previos (USANDO RUTAS)
    archivos_requeridos = [
        RUTAS['historico_csv'],  # baloto_historico_completo.csv en data/
        RUTAS['analisis_json']   # analisis_baloto.json en reports/
    ]
    
    archivos_faltantes = []
    for archivo in archivos_requeridos:
        if not os.path.exists(archivo):
            archivos_faltantes.append(os.path.basename(archivo))
    
    if archivos_faltantes:
        print(f"⚠️  Faltan archivos requeridos: {', '.join(archivos_faltantes)}")
        print("   Ejecuta las celdas 2 y 3 primero")
        
        # Intentar con lo que tengamos
        if not os.path.exists(RUTAS['historico_csv']):
            print(f"❌ No se puede ejecutar aprendizaje sin datos históricos")
            print(f"   Ubicación esperada: {RUTAS['historico_csv']}")
            return False
    
    print("🚀 Ejecutando sistema de aprendizaje automático...")
    print(f"   Script: {RUTAS['learning']}")
    
    try:
        # Ejecutar el sistema de aprendizaje (USANDO RUTAS)
        resultado = subprocess.run(
            [sys.executable, RUTAS['learning']],
            capture_output=True,
            text=True,
            encoding='utf-8',
            timeout=90  # 1.5 minutos máximo
        )
        
        # Verificar resultado
        if resultado.returncode == 0:
            print("✅ Aprendizaje completado exitosamente")
            
            # Procesar y mostrar output relevante
            output = resultado.stdout
            mostrar_resumen_aprendizaje(output)
            
            # Verificar archivos generados (USANDO RUTAS)
            verificar_archivos_aprendizaje()
            
            return True
            
        else:
            print("⚠️  El aprendizaje encontró algunos errores")
            
            # Intentar extraer información útil del output
            if resultado.stdout:
                lineas = resultado.stdout.split('\n')
                info_util = [l for l in lineas if any(
                    palabra in l for palabra in [
                        "INSIGHTS", "RECOMENDACIONES", "NÚMEROS", 
                        "SUPERBALOTAS", "Conocimiento guardado"
                    ])]
                
                if info_util:
                    print("\n📊 Información obtenida:")
                    for linea in info_util[:5]:
                        print(f"   {linea.strip()}")
            
            # Mostrar primeros errores
            if resultado.stderr:
                errores = [e.strip() for e in resultado.stderr.split('\n') if e.strip()]
                if errores:
                    print(f"\n⚠️  Errores (primeros 3):")
                    for error in errores[:3]:
                        print(f"   • {error}")
            
            # Verificar si al menos se generó conocimiento (USANDO RUTAS)
            if os.path.exists(RUTAS['knowledge_json']):
                print(f"\n📁 Usando conocimiento existente")
                tamano = os.path.getsize(RUTAS['knowledge_json'])
                print(f"   Tamaño: {tamano:,} bytes")
                print(f"   Ubicación: {RUTAS['knowledge_json']}")
                return True
            else:
                print(f"❌ No se pudo generar conocimiento en {RUTAS['knowledge_json']}")
                return False
                
    except subprocess.TimeoutExpired:
        print("⏰ Tiempo excedido en aprendizaje")
        
        # Verificar si hay conocimiento previo (USANDO RUTAS)
        if os.path.exists(RUTAS['knowledge_json']):
            print("📁 Usando conocimiento existente")
            tamano = os.path.getsize(RUTAS['knowledge_json'])
            tiempo_mod = os.path.getmtime(RUTAS['knowledge_json'])
            fecha_mod = datetime.fromtimestamp(tiempo_mod)
            
            print(f"   Tamaño: {tamano:,} bytes")
            print(f"   Ubicación: {RUTAS['knowledge_json']}")
            print(f"   Última actualización: {fecha_mod.strftime('%Y-%m-%d %H:%M:%S')}")
            return True
        else:
            print(f"❌ No hay conocimiento previo disponible en {RUTAS['knowledge_json']}")
            return False
            
    except Exception as e:
        print(f"❌ Error inesperado: {e}")
        return False

def mostrar_resumen_aprendizaje(output):
    """Muestra un resumen del output del aprendizaje"""
    lineas = output.split('\n')
    
    # Buscar secciones clave
    secciones_clave = [
        "INSIGHTS DEL SISTEMA",
        "RECOMENDACIONES ACTUALES",
        "NÚMEROS MÁS FRECUENTES",
        "SUPERBALOTAS MÁS FRECUENTES",
        "RESUMEN FINAL"
    ]
    
    print("\n📋 RESUMEN DEL APRENDIZAJE:")
    
    imprimiendo = False
    
    for linea in lineas:
        linea_stripped = linea.strip()
        
        # Detectar inicio de sección importante
        for seccion in secciones_clave:
            if seccion in linea:
                imprimiendo = True
                if linea_stripped:
                    print(f"\n{linea_stripped}")
                    print("-" * min(50, len(linea_stripped)))
                break
        
        # Imprimir contenido de sección si estamos en modo impresión
        if imprimiendo and linea_stripped:
            # Saltar líneas de separación largas
            if "="*40 in linea or "-"*40 in linea:
                continue
            
            # Formatear líneas de contenido
            if linea_stripped.startswith("•") or linea_stripped.startswith("   •"):
                print(f"   {linea_stripped}")
            elif any(marcador in linea for marcador in [
                "Número", "Superbalota", "Estrategia", "Confianza", "Considera"
            ]):
                print(f"   {linea_stripped}")
        
        # Detectar fin de sección (línea vacía después de contenido)
        elif imprimiendo and not linea_stripped:
            # Línea vacía podría indicar fin de sección
            # Esperar 2 líneas vacías consecutivas para mayor seguridad
            imprimiendo = False

def verificar_archivos_aprendizaje():
    """Verifica y muestra información de archivos generados (USANDO RUTAS)"""
    archivos_generados = [
        ("Conocimiento", RUTAS['knowledge_json']),
        ("Recomendaciones", RUTAS['recomendaciones_json']),
        ("Reporte", RUTAS['performance_report'])
    ]
    
    print("\n💾 ARCHIVOS GENERADOS:")
    
    for nombre, archivo in archivos_generados:
        if os.path.exists(archivo):
            tamano = os.path.getsize(archivo)
            tiempo_mod = os.path.getmtime(archivo)
            fecha_mod = datetime.fromtimestamp(tiempo_mod)
            
            if tamano > 1024*1024:
                tamano_str = f"{tamano/(1024*1024):.1f} MB"
            elif tamano > 1024:
                tamano_str = f"{tamano/1024:.1f} KB"
            else:
                tamano_str = f"{tamano} B"
            
            nombre_base = os.path.basename(archivo)
            print(f"   ✅ {nombre:20} {nombre_base:25} {tamano_str:>10} ({fecha_mod.strftime('%H:%M:%S')})")
            
            # Mostrar contenido clave si es pequeño y es recomendaciones
            if "recomendaciones" in archivo.lower() and tamano < 5000:
                try:
                    with open(archivo, "r") as f:
                        datos = json.load(f)
                    
                    if 'strategy' in datos:
                        print(f"     • Estrategia: {datos['strategy'].upper()}")
                    if 'numbers' in datos and datos['numbers']:
                        print(f"     • Números: {datos['numbers'][:5]}")
                    if 'superballots' in datos and datos['superballots']:
                        print(f"     • Superbalotas: {datos['superballots']}")
                        
                except:
                    pass
        else:
            nombre_base = os.path.basename(archivo)
            print(f"   ❌ {nombre:20} {nombre_base:25} No generado")

def crear_archivo_aprendizaje_basico():
    """Crea un archivo básico de aprendizaje si no existe (USANDO RUTAS)"""
    # Crear la carpeta scripts si no existe
    os.makedirs(os.path.dirname(RUTAS['learning']), exist_ok=True)
    
    codigo_basico = f'''
"""
learning_system_enhanced.py - Sistema de aprendizaje básico
Versión simplificada para uso inmediato
"""

import json
import pandas as pd
from collections import Counter
from datetime import datetime

def main():
    print("🧠 SISTEMA DE APRENDIZAJE BÁSICO")
    
    try:
        # Cargar datos (USANDO RUTAS RELATIVAS)
        df = pd.read_csv("{RUTAS['historico_csv']}")
        baloto_df = df[df['tipo'] == 'Baloto']
        
        print(f"📊 Analizando {{len(baloto_df)}} sorteos...")
        
        # Análisis simple
        all_numbers = []
        for i in range(1, 6):
            all_numbers.extend(baloto_df[f'num{{i}}'].tolist())
        
        number_freq = Counter(all_numbers)
        top_numbers = number_freq.most_common(10)
        
        sb_freq = Counter(baloto_df['num6'].tolist())
        top_sb = sb_freq.most_common(5)
        
        # Crear recomendaciones
        recomendaciones = {{
            'strategy': 'caliente',
            'numbers': [num for num, _ in top_numbers[:8]],
            'superballots': [sb for sb, _ in top_sb[:3]],
            'confidence': 0.6,
            'timestamp': datetime.now().isoformat(),
            'data_points': len(baloto_df)
        }}
        
        # Guardar en reports/ (USANDO RUTAS RELATIVAS)
        import os
        os.makedirs(os.path.dirname("{RUTAS['recomendaciones_json']}"), exist_ok=True)
        
        with open("{RUTAS['recomendaciones_json']}", "w") as f:
            json.dump(recomendaciones, f, indent=2)
        
        print(f"✅ Recomendaciones generadas")
        print(f"   • Estrategia: {{recomendaciones['strategy'].upper()}}")
        print(f"   • Números: {{recomendaciones['numbers'][:5]}}")
        print(f"   • Guardado en: {{"{RUTAS['recomendaciones_json']}"}}")
        
    except Exception as e:
        print(f"❌ Error: {{e}}")

if __name__ == "__main__":
    main()
'''
    
    try:
        with open(RUTAS['learning'], "w", encoding='utf-8') as f:
            f.write(codigo_basico)
        print(f"✅ Archivo básico de aprendizaje creado en: {RUTAS['learning']}")
        return True
    except Exception as e:
        print(f"❌ Error creando archivo: {e}")
        return False

# Ejecutar aprendizaje automático
print("\n🔍 Iniciando proceso de aprendizaje automático...")
ejecutar_aprendizaje_automatico()

print("\n" + "=" * 70)
```

```python
# ============================================================================
# CELDA 6: GENERADOR DE JUEGOS - VERSIÓN MEJORADA PARA 4 ESTRATEGIAS
# ============================================================================

print("🎲 GENERADOR DE JUEGOS - 4 ESTRATEGIAS SIMULTÁNEAS")
print("=" * 70)

def generar_cuatro_estrategias_una_fecha(fecha):
    """Genera 4 juegos, uno por cada estrategia, para una misma fecha"""
    
    print(f"\n🎯 GENERANDO 4 ESTRATEGIAS PARA: {fecha}")
    print("-" * 50)
    
    estrategias = ["caliente", "fria", "balanceada", "mixta"]
    resultados = []
    
    for i, estrategia in enumerate(estrategias, 1):
        print(f"\n🔢 Estrategia {i}/4: {estrategia.upper()}")
        
        try:
            # Ejecutar new_game.py para esta estrategia
            resultado = subprocess.run(
                [sys.executable, RUTAS['new_game'], fecha, "--estrategia", estrategia],
                capture_output=True,
                text=True,
                encoding='utf-8',
                timeout=15  # 15 segundos por estrategia
            )
            
            if resultado.returncode == 0:
                print(f"   ✅ Generado exitosamente")
                
                # Extraer números del output para mostrar
                for linea in resultado.stdout.split('\n'):
                    if "NÚMEROS PRINCIPALES:" in linea:
                        print(f"   {linea.strip()}")
                    elif "SUPERBALOTA:" in linea:
                        print(f"   {linea.strip()}")
                    elif "Puntaje:" in linea:
                        print(f"   {linea.strip()}")
                
                resultados.append((estrategia, True, ""))
            else:
                error_msg = "Error desconocido"
                if resultado.stderr:
                    errores = [e.strip() for e in resultado.stderr.split('\n') if e.strip()]
                    if errores:
                        error_msg = errores[0][:50]
                
                print(f"   ❌ Error: {error_msg}")
                resultados.append((estrategia, False, error_msg))
                
        except subprocess.TimeoutExpired:
            print(f"   ⏰ Tiempo excedido")
            resultados.append((estrategia, False, "Timeout"))
        except Exception as e:
            error_msg = str(e)[:50]
            print(f"   ⚠️  Error: {error_msg}")
            resultados.append((estrategia, False, error_msg))
    
    # Resumen
    exitosos = sum(1 for r in resultados if r[1])
    print(f"\n📊 RESUMEN: {exitosos}/4 estrategias generadas para {fecha}")
    
    return resultados

def generar_cuatro_estrategias_varias_fechas(fechas):
    """Genera 4 estrategias para cada fecha en la lista"""
    
    print(f"\n🎯 GENERANDO 4 ESTRATEGIAS PARA {len(fechas)} FECHAS")
    print("=" * 50)
    
    total_generados = 0
    total_intentos = len(fechas) * 4
    todos_resultados = []
    
    for fecha_idx, fecha in enumerate(fechas, 1):
        print(f"\n📅 FECHA {fecha_idx}/{len(fechas)}: {fecha}")
        print("-" * 40)
        
        resultados_fecha = generar_cuatro_estrategias_una_fecha(fecha)
        todos_resultados.append((fecha, resultados_fecha))
        
        # Contar exitosos de esta fecha
        exitosos_fecha = sum(1 for r in resultados_fecha if r[1])
        total_generados += exitosos_fecha
        
        print(f"✅ {exitosos_fecha}/4 generados para {fecha}")
    
    # Resumen final
    print(f"\n" + "="*70)
    print(f"📈 RESUMEN FINAL")
    print(f"="*70)
    print(f"   • Total fechas: {len(fechas)}")
    print(f"   • Total juegos intentados: {total_intentos}")
    print(f"   • Total juegos generados: {total_generados}")
    print(f"   • Tasa de éxito: {total_generados/total_intentos*100:.1f}%")
    
    # Mostrar distribución por estrategia
    if total_generados > 0:
        print(f"\n🎯 DISTRIBUCIÓN POR ESTRATEGIA:")
        estrategias = ["caliente", "fria", "balanceada", "mixta"]
        for estrategia in estrategias:
            exitosos_estrategia = sum(1 for fecha, resultados in todos_resultados 
                                     for r in resultados if r[0] == estrategia and r[1])
            print(f"   • {estrategia.upper():12}: {exitosos_estrategia}/{len(fechas)} fechas")
    
    print(f"\n💾 Todos los juegos guardados en: {RUTAS['mis_juegos_csv']}")
    
    return todos_resultados

def generar_estrategias_para_proximo_mes():
    """Genera 4 estrategias para todos los sorteos del próximo mes"""
    
    from datetime import datetime, timedelta
    
    print(f"\n📅 GENERANDO PARA PRÓXIMO MES")
    print("=" * 50)
    
    # Calcular fechas de los próximos sorteos (martes y viernes/sábado)
    hoy = datetime.now()
    fechas = []
    
    # Próximos 30 días
    for i in range(1, 31):
        fecha_futura = hoy + timedelta(days=i)
        
        # Baloto normalmente sale martes y viernes/sábado
        # Ajusta según el horario real de sorteos en tu país
        if fecha_futura.weekday() in [1, 4, 5]:  # Martes, Viernes, Sábado
            fecha_str = fecha_futura.strftime("%d-%m-%Y")
            fechas.append(fecha_str)
    
    print(f"📅 Encontradas {len(fechas)} fechas de sorteos en próximos 30 días:")
    for i, fecha in enumerate(fechas[:5], 1):
        print(f"   {i}. {fecha}")
    if len(fechas) > 5:
        print(f"   ... y {len(fechas)-5} más")
    
    if len(fechas) > 0:
        confirmar = input(f"\n¿Generar 4 estrategias para {len(fechas)} fechas? (s/n): ")
        if confirmar.lower() == 's':
            return generar_cuatro_estrategias_varias_fechas(fechas)
        else:
            print("❌ Generación cancelada")
            return None
    else:
        print("❌ No se encontraron fechas de sorteos en el próximo mes")
        return None

# ============================================================================
# CONFIGURACIÓN MEJORADA
# ============================================================================

# OPCIÓN 1: Generar 4 estrategias para UNA fecha
GENERAR_CUATRO_UNA_FECHA = True  # Cambia a True para usar
FECHA_UNICA = "10-01-2026"      # Fecha futura para jugar

# OPCIÓN 2: Generar 4 estrategias para MÚLTIPLES fechas
GENERAR_CUATRO_MULTIPLES_FECHAS = False  # Cambia a True para usar
FECHAS_MULTIPLES = ["05-01-2026", "06-01-2026", "07-01-2026", "08-01-2026"]

# OPCIÓN 3: Generar para próximo mes automáticamente
GENERAR_PARA_PROXIMO_MES = False  # Cambia a True para usar

# OPCIÓN 4: Ver juegos existentes (siempre útil)
VER_JUEGOS_EXISTENTES = True
LIMITE_VISUALIZACION = 12  # Mostrar más porque habrá 4 por fecha

# ============================================================================
# EJECUCIÓN (no modifiques esto)
# ============================================================================

# Importar sys si no está
import sys

# 1. Generar 4 estrategias para una fecha
if GENERAR_CUATRO_UNA_FECHA:
    print("\n" + "="*70)
    print("🎯 OPCIÓN 1: 4 ESTRATEGIAS PARA UNA FECHA")
    print("="*70)
    generar_cuatro_estrategias_una_fecha(FECHA_UNICA)

# 2. Generar 4 estrategias para múltiples fechas
if GENERAR_CUATRO_MULTIPLES_FECHAS:
    print("\n" + "="*70)
    print("🎯 OPCIÓN 2: 4 ESTRATEGIAS PARA MÚLTIPLES FECHAS")
    print("="*70)
    generar_cuatro_estrategias_varias_fechas(FECHAS_MULTIPLES)

# 3. Generar para próximo mes
if GENERAR_PARA_PROXIMO_MES:
    print("\n" + "="*70)
    print("🎯 OPCIÓN 3: 4 ESTRATEGIAS PARA PRÓXIMO MES")
    print("="*70)
    generar_estrategias_para_proximo_mes()

# 4. Ver juegos existentes
if VER_JUEGOS_EXISTENTES:
    print("\n" + "="*70)
    print("📋 JUEGOS EXISTENTES (últimos 15)")
    print("="*70)
    
    if os.path.exists(RUTAS['mis_juegos_csv']):
        try:
            df = pd.read_csv(RUTAS['mis_juegos_csv'])
            if len(df) > 0:
                print(f"📊 Total juegos en archivo: {len(df)}")
                
                # Mostrar agrupados por fecha
                fechas_unicas = df['fecha_juego'].unique()
                print(f"📅 Fechas con juegos: {len(fechas_unicas)}")
                
                # Mostrar últimos juegos
                print(f"\n📝 ÚLTIMOS {min(LIMITE_VISUALIZACION, len(df))} JUEGOS:")
                for idx, row in df.head(LIMITE_VISUALIZACION).iterrows():
                    fecha = row.get('fecha_juego', 'N/A')
                    nums = [str(int(row[f'num{i}'])) if pd.notna(row.get(f'num{i}')) else '' for i in range(1, 6)]
                    sb = str(int(row['num6'])) if pd.notna(row.get('num6')) else ''
                    estrategia = row.get('estrategia', 'N/A')
                    
                    print(f"\n   📅 {fecha}")
                    print(f"      🎯 {estrategia.upper():12} | SB: {sb}")
                    print(f"      🔢 {' - '.join(nums)}")
        except Exception as e:
            print(f"❌ Error leyendo juegos: {e}")
    else:
        print("📭 No hay juegos generados aún")

print("\n" + "=" * 70)
print("✅ PROCESO COMPLETADO")
print("=" * 70)
```
